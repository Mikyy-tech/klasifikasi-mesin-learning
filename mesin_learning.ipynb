```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

```python
from enum import KEEP
iris = load_iris(as_frame=True)
keep = iris.target.isin([1, 2])
X = iris.data.loc[keep,
    ['petal length (cm)', 'petal width (cm)']]
y = (iris.target.loc[keep] == 2).astype(int)
print(X.shape, y.value_counts().to_dict())
```

```python
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.20,
    random_state=42, stratify=y
)
print(len(y_tr), len(y_te))
```

```python
nb = GaussianNB()
nb.fit(X_tr, y_tr)
yhat_nb = nb.predict(X_te)
p_nb = nb.predict_proba(X_te)[:, 1]
print(yhat_nb [:5], p_nb[:5])
```

```python
lr = Pipeline([
    ('scaler', StandardScaler()),
    ('logistic', LogisticRegression(max_iter=1000, random_state=42))
])
lr.fit(X_tr, y_tr)
yhat_lr = lr.predict(X_te)
P_lr = lr.predict_proba(X_te)[:, 1]
print(yhat_lr[:5], P_lr[:5])
```

```python
from sklearn.metrics import (accuracy_score,
    precision_score, recall_score, f1_score)
for name,pred in [('NB', yhat_nb), ('LR', yhat_lr)]:
    print(name)
    print('accuracy:', accuracy_score(y_te, pred))
```
